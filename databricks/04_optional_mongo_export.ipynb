# 04 (alt) - Mongo Export

```python
from pymongo import MongoClient
mongo_user=dbutils.secrets.get("scope","mongo_user")
mongo_pwd=dbutils.secrets.get("scope","mongo_pwd")
mongo_host=dbutils.secrets.get("scope","mongo_host")
uri=f"mongodb+srv://{mongo_user}:{mongo_pwd}@{mongo_host}/?retryWrites=true&w=majority"
client=MongoClient(uri)
db=client["olist"]
coll=db["daily_orders"]
data=[r.asDict() for r in spark.table("gold.daily_orders").collect()] 
if data: coll.delete_many({}); coll.insert_many(data)
print("Exported to Mongo")
```
