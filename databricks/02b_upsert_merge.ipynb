# 02b - MERGE Upsert

```python
dbutils.widgets.text("curated_base","abfss://olistdata@olistecommdatastorage.dfs.core.windows.net/curated"); CUR=dbutils.widgets.get("curated_base").rstrip("/")
```

```python
src=spark.read.format("delta").load(f"{CUR}/bronze/orders_incremental"); tgt=f"{CUR}/silver/orders"; src.createOrReplaceTempView("src"); spark.sql(f"""MERGE INTO delta.`{tgt}` AS t USING src AS s ON t.order_id=s.order_id WHEN MATCHED THEN UPDATE SET * WHEN NOT MATCHED THEN INSERT *"""); print("MERGE completed")
```
