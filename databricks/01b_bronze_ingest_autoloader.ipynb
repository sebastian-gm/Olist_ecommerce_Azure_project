# 01b - Bronze Ingest (Auto Loader -- Incremantal)

```python
dbutils.widgets.text("raw_inbox","abfss://olistdata@olistecommdatastorage.dfs.core.windows.net/inbox"); dbutils.widgets.text("curated_base","abfss://olistdata@olistecommdatastorage.dfs.core.windows.net/curated"); INBOX=dbutils.widgets.get("raw_inbox").rstrip("/"); CUR=dbutils.widgets.get("curated_base").rstrip("/")
```

```python
(spark.readStream.format("cloudFiles").option("cloudFiles.format","csv").option("header","true").option("cloudFiles.inferColumnTypes","true").option("cloudFiles.schemaLocation",f"{CUR}/_schemas/orders").load(f"{INBOX}/orders/").writeStream.format("delta").outputMode("append").trigger(availableNow=True).option("checkpointLocation",f"{CUR}/_checkpoints/orders").start(f"{CUR}/bronze/orders_incremental"))
```
