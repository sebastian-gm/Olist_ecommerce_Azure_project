# 06 - Delta Maintenance

```python
dbutils.widgets.text("curated_base","abfss://olistdata@olistecommdatastorage.dfs.core.windows.net/curated")
CUR=dbutils.widgets.get("curated_base").rstrip("/")
```

```python
spark.sql(f"OPTIMIZE delta.`{CUR}/silver/orders` ZORDER BY (order_date, customer_id)")
spark.sql(f"VACUUM delta.`{CUR}/silver/orders` RETAIN 168 HOURS")
print("Maintenance done")
```
