# 03 - Gold Marts (daily_orders)

```python
dbutils.widgets.text("curated_base", "abfss://olistdata@olistecommdatastorage.dfs.core.windows.net/curated")
CUR = dbutils.widgets.get("curated_base").rstrip("/")
```

```python
spark.sql(f"CREATE DATABASE IF NOT EXISTS gold LOCATION '{CUR}/gold'")
silver_orders = f"{CUR}/silver/orders"
spark.sql(f'''
CREATE OR REPLACE TABLE gold.daily_orders USING DELTA AS
SELECT order_date, COUNT(*) AS orders
FROM delta.`{silver_orders}`
GROUP BY order_date
''')
print("Gold created: gold.daily_orders")
spark.table("gold.daily_orders").show(10, truncate=False)
```
