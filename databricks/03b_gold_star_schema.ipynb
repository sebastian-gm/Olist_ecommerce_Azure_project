# 03b - Gold Star Schema

```python
dbutils.widgets.text("curated_base","abfss://olistdata@olistecommdatastorage.dfs.core.windows.net/curated"); CUR=dbutils.widgets.get("curated_base").rstrip("/")
```

```python
spark.sql(f"CREATE DATABASE IF NOT EXISTS gold LOCATION '{CUR}/gold'")
spark.sql(f"""CREATE OR REPLACE TABLE gold.dim_customer USING DELTA AS SELECT DISTINCT customer_id FROM delta.`{CUR}/silver/customers`""")
spark.sql(f"""CREATE OR REPLACE TABLE gold.dim_product USING DELTA AS SELECT DISTINCT product_id, product_category_name FROM delta.`{CUR}/silver/products`""")
spark.sql(f"""CREATE OR REPLACE TABLE gold.fact_order_items USING DELTA AS SELECT oi.order_id, oi.product_id, oi.seller_id, CAST(oi.price AS DOUBLE) AS price, CAST(oi.freight_value AS DOUBLE) AS freight_value FROM delta.`{CUR}/silver/order_items` oi""")
print("Star schema created")
```
